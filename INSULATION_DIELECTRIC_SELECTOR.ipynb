```python
# ============================================================
# INSULATION / DIELECTRIC SELECTOR
# Materials Science Assignment
# ============================================================

# Material database
materials = {
    "PVC": {
        "strength": 40,      # kV/mm
        "temperature": 70    # °C
    },
    "XLPE": {
        "strength": 30,
        "temperature": 90
    },
    "PTFE": {
        "strength": 60,
        "temperature": 260
    },
    "Epoxy": {
        "strength": 20,
        "temperature": 130
    },
    "Polyimide": {
        "strength": 200,
        "temperature": 400
    }
}

print("=" * 60)
print("         INSULATION / DIELECTRIC SELECTOR")
print("=" * 60)

# -----------------------------
# User Input
# -----------------------------
while True:
    try:
        voltage = float(input("Enter Voltage Rating (V): "))
        if voltage <= 0:
            print("Voltage must be greater than zero.\n")
            continue
        break
    except ValueError:
        print("Please enter a valid number.\n")

while True:
    try:
        temperature = float(input("Enter Maximum Operating Temperature (°C): "))
        break
    except ValueError:
        print("Please enter a valid number.\n")

while True:
    try:
        thickness = float(input("Enter Available Insulation Thickness (mm): "))
        if thickness <= 0:
            print("Thickness must be greater than zero.\n")
            continue
        break
    except ValueError:
        print("Please enter a valid number.\n")

# --------------------------------------------------
# Engineering Calculation
# --------------------------------------------------
required_strength = voltage / thickness / 1000  # convert V/mm to kV/mm

print("\n" + "=" * 60)
print("CALCULATION")
print("=" * 60)

print(f"Required Dielectric Strength : {required_strength:.2f} kV/mm")
print(f"Operating Temperature        : {temperature:.1f} °C")

print("\n" + "=" * 60)
print("MATERIAL EVALUATION")
print("=" * 60)

passing_materials = []

for material, data in materials.items():

    strength_ok = data["strength"] >= required_strength
    temp_ok = data["temperature"] >= temperature

    print(f"\nMaterial: {material}")
    print(f"Dielectric Strength : {data['strength']} kV/mm")
    print(f"Maximum Temperature : {data['temperature']} °C")

    if strength_ok and temp_ok:

        print("Status : PASS")
        print("Reason : Meets both dielectric strength and temperature requirements.")

        passing_materials.append({
            "name": material,
            "strength": data["strength"],
            "temperature": data["temperature"]
        })

    else:

        print("Status : FAIL")
        print("Reason(s):")

        if not strength_ok:
            print(
                f"  • Dielectric strength is too low "
                f"({data['strength']} < {required_strength:.2f} kV/mm)"
            )

        if not temp_ok:
            print(
                f"  • Maximum temperature is too low "
                f"({data['temperature']}°C < {temperature:.1f}°C)"
            )

print("\n" + "=" * 60)
print("FINAL RECOMMENDATION")
print("=" * 60)

if len(passing_materials) == 0:

    print("No available insulation material satisfies the given requirements.")

else:

    # Recommend the material with the highest dielectric strength
    best = max(passing_materials, key=lambda x: x["strength"])

    print(f"Recommended Material : {best['name']}")
    print(f"Dielectric Strength  : {best['strength']} kV/mm")
    print(f"Maximum Temperature  : {best['temperature']} °C")

    # Safety margin
    safety_factor = best["strength"] / required_strength

    print(f"Safety Factor        : {safety_factor:.2f}")

    if safety_factor >= 2:
        print("Engineering Assessment:")
        print("This material provides a comfortable safety margin.")
    else:
        print("Engineering Assessment:")
        print("The material meets the requirements, but the safety margin is relatively small.")

print("\nProgram Finished.")
```
